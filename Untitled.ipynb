```python
import random
import ipaddress
import pandas as pd

# -----------------------------
# SETTINGS
# -----------------------------
random.seed(42)

NORMAL_COUNT = 600
ERROR_COUNT_PER_TYPE = 80

NETWORKS = ["HospitalNet", "SchoolNet", "CampusNet", "OfficeNet"]
DEVICES = ["R1", "R2", "R3", "R4", "R5", "R6"]

ERROR_TYPES = [
    "duplicate_ip",
    "subnet_mask_mismatch",
    "network_mismatch",
    "status_mismatch",
    "invalid_host_ip"
]

# -----------------------------
# HELPER FUNCTIONS
# -----------------------------
def get_prefix(mask):
    return ipaddress.IPv4Network(f"0.0.0.0/{mask}").prefixlen

def get_network_address(ip, mask):
    return str(ipaddress.ip_network(f"{ip}/{mask}", strict=False).network_address)

def is_valid_host_ip(ip, mask):
    network = ipaddress.ip_network(f"{ip}/{mask}", strict=False)
    ip_obj = ipaddress.ip_address(ip)
    return ip_obj != network.network_address and ip_obj != network.broadcast_address

def create_normal_sample(sample_id):
    network = random.choice(NETWORKS)
    device_a, device_b = random.sample(DEVICES, 2)

    subnet_id = random.randint(1, 220)

    ip_a = f"10.{subnet_id}.0.1"
    ip_b = f"10.{subnet_id}.0.2"

    return {
        "Sample_ID": sample_id,
        "Network": network,
        "Link_ID": f"L{sample_id}",
        "Device_A": device_a,
        "Device_B": device_b,
        "Interface_A": "Fa0/1",
        "Interface_B": "Fa0/1",
        "IP_A": ip_a,
        "IP_B": ip_b,
        "Subnet_A": "255.255.255.252",
        "Subnet_B": "255.255.255.252",
        "Status_A": "up",
        "Status_B": "up",
        "Label": 0,
        "Risk_Level": "Low",
        "Error_Type": "none"
    }

def create_error_sample(sample_id, error_type):
    row = create_normal_sample(sample_id)

    row["Label"] = 1
    row["Error_Type"] = error_type

    if error_type == "duplicate_ip":
        row["IP_B"] = row["IP_A"]
        row["Risk_Level"] = "High"

    elif error_type == "subnet_mask_mismatch":
        row["Subnet_B"] = "255.255.255.0"
        row["Risk_Level"] = "Medium"

    elif error_type == "network_mismatch":
        row["IP_B"] = f"192.168.{random.randint(1, 220)}.1"
        row["Risk_Level"] = "High"

    elif error_type == "status_mismatch":
        row["Status_B"] = "down"
        row["Risk_Level"] = "Medium"

    elif error_type == "invalid_host_ip":
        # With /30 network, .0 is network address, not valid host
        subnet_id = random.randint(1, 220)
        row["IP_A"] = f"10.{subnet_id}.0.0"
        row["IP_B"] = f"10.{subnet_id}.0.2"
        row["Risk_Level"] = "High"

    return row

def generate_features(raw_df):
    feature_rows = []

    for _, row in raw_df.iterrows():
        ip_a = row["IP_A"]
        ip_b = row["IP_B"]
        subnet_a = row["Subnet_A"]
        subnet_b = row["Subnet_B"]

        prefix_a = get_prefix(subnet_a)
        prefix_b = get_prefix(subnet_b)

        duplicate_ip = int(ip_a == ip_b)
        same_subnet_mask = int(subnet_a == subnet_b)
        same_network = int(
            get_network_address(ip_a, subnet_a) == get_network_address(ip_b, subnet_b)
        )
        prefix_difference = abs(prefix_a - prefix_b)
        status_consistency = int(row["Status_A"] == row["Status_B"])
        both_interfaces_up = int(row["Status_A"] == "up" and row["Status_B"] == "up")

        invalid_host_a = int(not is_valid_host_ip(ip_a, subnet_a))
        invalid_host_b = int(not is_valid_host_ip(ip_b, subnet_b))

        feature_rows.append({
            "Sample_ID": row["Sample_ID"],
            "Network": row["Network"],
            "Link_ID": row["Link_ID"],
            "Device_A": row["Device_A"],
            "Device_B": row["Device_B"],

            # Generated ML features
            "duplicate_ip": duplicate_ip,
            "same_subnet_mask": same_subnet_mask,
            "same_network": same_network,
            "prefix_difference": prefix_difference,
            "status_consistency": status_consistency,
            "both_interfaces_up": both_interfaces_up,
            "invalid_host_a": invalid_host_a,
            "invalid_host_b": invalid_host_b,

            # Target/output columns
            "Label": row["Label"],
            "Risk_Level": row["Risk_Level"],
            "Error_Type": row["Error_Type"]
        })

    return pd.DataFrame(feature_rows)

# -----------------------------
# GENERATE RAW DATASET
# -----------------------------
rows = []
sample_id = 1

# Normal samples
for _ in range(NORMAL_COUNT):
    rows.append(create_normal_sample(sample_id))
    sample_id += 1

# Risky samples
for error_type in ERROR_TYPES:
    for _ in range(ERROR_COUNT_PER_TYPE):
        rows.append(create_error_sample(sample_id, error_type))
        sample_id += 1

raw_df = pd.DataFrame(rows)

# Shuffle rows
raw_df = raw_df.sample(frac=1, random_state=42).reset_index(drop=True)

# Generate ML feature dataset
features_df = generate_features(raw_df)

# -----------------------------
# SAVE FILES
# -----------------------------
raw_df.to_csv("dataset_raw.csv", index=False)
features_df.to_csv("dataset_features.csv", index=False)

# Optional Excel file
with pd.ExcelWriter("dataset_final.xlsx", engine="openpyxl") as writer:
    raw_df.to_excel(writer, sheet_name="Raw_Dataset", index=False)
    features_df.to_excel(writer, sheet_name="Feature_Dataset", index=False)

# -----------------------------
# PRINT SUMMARY
# -----------------------------
print("Dataset created successfully!")
print("\nRaw dataset shape:", raw_df.shape)
print("Feature dataset shape:", features_df.shape)

print("\nLabel distribution:")
print(raw_df["Label"].value_counts())

print("\nRisk level distribution:")
print(raw_df["Risk_Level"].value_counts())

print("\nError type distribution:")
print(raw_df["Error_Type"].value_counts())

print("\nFiles created:")
print("1. dataset_raw.csv")
print("2. dataset_features.csv")
print("3. dataset_final.xlsx")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load datasets
raw_df = pd.read_csv("dataset_raw.csv")
feature_df = pd.read_csv("dataset_features.csv")

# -----------------------------
# 1. Basic Dataset Information
# -----------------------------
print("Raw Dataset Shape:", raw_df.shape)
print("Feature Dataset Shape:", feature_df.shape)

print("\nRaw Dataset Columns:")
print(raw_df.columns.tolist())

print("\nFeature Dataset Columns:")
print(feature_df.columns.tolist())

print("\nFirst 5 rows of raw dataset:")
print(raw_df.head())

print("\nFirst 5 rows of feature dataset:")
print(feature_df.head())

# -----------------------------
# 2. Missing Values Check
# -----------------------------
print("\nMissing values in raw dataset:")
print(raw_df.isnull().sum())

print("\nMissing values in feature dataset:")
print(feature_df.isnull().sum())

# -----------------------------
# 3. Duplicate Rows Check
# -----------------------------
print("\nDuplicate rows in raw dataset:", raw_df.duplicated().sum())
print("Duplicate rows in feature dataset:", feature_df.duplicated().sum())

# -----------------------------
# 4. Label Distribution
# -----------------------------
print("\nLabel distribution:")
print(raw_df["Label"].value_counts())

plt.figure(figsize=(6,4))
raw_df["Label"].value_counts().plot(kind="bar")
plt.title("Label Distribution")
plt.xlabel("Label (0 = Normal, 1 = Risky)")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# -----------------------------
# 5. Risk Level Distribution
# -----------------------------
print("\nRisk level distribution:")
print(raw_df["Risk_Level"].value_counts())

plt.figure(figsize=(6,4))
raw_df["Risk_Level"].value_counts().plot(kind="bar")
plt.title("Risk Level Distribution")
plt.xlabel("Risk Level")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# -----------------------------
# 6. Error Type Distribution
# -----------------------------
print("\nError type distribution:")
print(raw_df["Error_Type"].value_counts())

plt.figure(figsize=(8,4))
raw_df["Error_Type"].value_counts().plot(kind="bar")
plt.title("Error Type Distribution")
plt.xlabel("Error Type")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# -----------------------------
# 7. Network Distribution
# -----------------------------
print("\nNetwork distribution:")
print(raw_df["Network"].value_counts())

plt.figure(figsize=(6,4))
raw_df["Network"].value_counts().plot(kind="bar")
plt.title("Network Distribution")
plt.xlabel("Network")
plt.ylabel("Count")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# -----------------------------
# 8. Feature Summary
# -----------------------------
feature_columns = [
    "duplicate_ip",
    "same_subnet_mask",
    "same_network",
    "prefix_difference",
    "status_consistency",
    "both_interfaces_up",
    "invalid_host_a",
    "invalid_host_b"
]

print("\nFeature summary:")
print(feature_df[feature_columns].describe())

# -----------------------------
# 9. Feature vs Label Analysis
# -----------------------------
for col in feature_columns:
    print(f"\n{col} vs Label:")
    print(pd.crosstab(feature_df[col], feature_df["Label"]))

# -----------------------------
# 10. Correlation Matrix
# -----------------------------
corr = feature_df[feature_columns + ["Label"]].corr()

print("\nCorrelation with Label:")
print(corr["Label"].sort_values(ascending=False))

plt.figure(figsize=(8,6))
plt.imshow(corr, aspect="auto")
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```
